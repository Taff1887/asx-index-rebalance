# 02 — FMP vs Yahoo validation

Reconciles the two raw panels and writes the data-quality reports to `outputs/`.
Discrepancies are tagged with a severity bucket (low/medium/high) and any
missing observations are flagged before the price panel is reconciled.


```python
import pandas as pd
from asxrebalance.paths import RAW_FMP_DIR, RAW_YAHOO_DIR
from asxrebalance.data.validation import generate_data_quality_report

def _load(dir_):
    frames = []
    for f in sorted(dir_.glob('*.csv')):
        if f.stem in ('shares', 'iwf'): continue
        df = pd.read_csv(f, parse_dates=['date']); df['ticker'] = f.stem; frames.append(df)
    return pd.concat(frames, ignore_index=True)

fmp = _load(RAW_FMP_DIR); yahoo = _load(RAW_YAHOO_DIR)
report = generate_data_quality_report(fmp, yahoo)
report['summary']
```
